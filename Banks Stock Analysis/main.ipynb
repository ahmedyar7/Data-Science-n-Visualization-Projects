```python
import pandas as pd
from pandas_datareader import data, wb
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime as dt
import plotly.express as px
import cufflinks as cf

cf.go_offline()
```

```python
start = dt(2006, 1, 1)
end = dt(2016, 1, 1)
```

```python
BAC = data.DataReader("BAC", "stooq", start, end)  # Bank Of America
BAC.head()
```

```python
C = data.DataReader("C", "stooq", start, end)  # CitiGroup Bank
C.head()
```

```python
MS = data.DataReader("MS", "stooq", start, end)  # Morgan Stanly Bank
MS.head()
```

```python
JPM = data.DataReader("JPM", "stooq", start, end)  # JP Morgan Bank
JPM.head()
```

```python
WFC = data.DataReader("WFC", "stooq", start, end)
WFC.head()
```

```python
GS = data.DataReader("GS", "stooq", start, end)
GS.head()
```

```python
bank_ticker = ["BAC", "C", "MS", "GS", "JPM", "WFC"]
```

```python
bank_stocks_df = pd.concat([BAC, C, MS, GS, JPM, WFC], axis=1, keys=bank_ticker)
bank_stocks_df.head()
```

```python
bank_stocks_df.columns.names = ["Bank_Ticker", "Stock_Info"]
bank_stocks_df.head()
```

```python
bank_stocks_df.xs(
    key="Close",
    axis=1,
    level="Stock_Info",
).max()
```

```python
return_df = pd.DataFrame()
```

```python
for tick in bank_ticker:
    return_df[tick + "_Returns"] = bank_stocks_df[tick]["Close"].pct_change(
        fill_method=None
    )

return_df.head()
```

```python
sns.pairplot(return_df)
```

```python
return_df.idxmax()
```

```python
return_df.idxmin()
```

```python
plt.figure(figsize=(12, 5), dpi=500)

plt.title("Return on JP Morgan Bank")
plt.xlabel("Stock Percentage")
plt.ylabel("Stock Count")

sns.histplot(
    data=return_df["2015-01-01":"2016-12-31"]["JPM_Returns"],
    bins=50,
    kde=True,
)

plt.show()
```

```python
plt.figure(figsize=(12, 5), dpi=500)

plt.title("Returns on CitiGroup Investment Bank")
plt.xlabel("Stock-Percentage")
plt.ylabel("Stock Count")

sns.histplot(
    data=return_df["2015-01-01":"2015-12-31"]["C_Returns"],
    color="orange",
    bins=50,
    kde=True,
)

plt.show()
```

```python
plt.figure(figsize=(12, 5), dpi=500)


plt.title("Closing Stock Price of Banks")
plt.xlabel("Date")
plt.ylabel("Stock Price")

for ticks in bank_ticker:
    bank_stocks_df[ticks]["Close"].plot(label=ticks)


plt.legend()
plt.show()
```

```python
bank_stocks_df.xs(key="Close", level="Stock_Info", axis=1).iplot()
```

```python
bac_sorted = BAC.sort_index()

rolling_avg = bac_sorted["Close"]["2008-01-01":"2009-01-01"].rolling(window=30).mean()
normal_bac = bac_sorted["Close"]["2008-01-01":"2009-01-01"]

plt.figure(figsize=(12, 5), dpi=500)

plt.title("Rolling Average Vs Normal Distribution")

plt.xlabel("Date")
plt.ylabel("Stock Count")

plt.plot(
    rolling_avg.index,
    rolling_avg.values,
    color="red",
    ls="-.",
    label="30 Day Rolling Average",
)

plt.plot(
    normal_bac.index,
    normal_bac.values,
    color="dodgerblue",
    marker=".",
    markerfacecolor="orange",
    markeredgecolor="orange",
    label="Normal Distribution",
)

plt.tight_layout()
plt.legend()


plt.show()
```
