```python
import pandas as pd
from pandas_datareader import data, wb
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime as dt
import plotly.express as px
import cufflinks as cf

cf.go_offline()
```

```python
start = dt(2006, 1, 1)
end = dt(2016, 1, 1)
```

```python
BAC = data.DataReader("BAC", "stooq", start, end)  # Bank Of America
BAC.head()
```

```python
C = data.DataReader("C", "stooq", start, end)  # CitiGroup Bank
C.head()
```

```python
MS = data.DataReader("MS", "stooq", start, end)  # Morgan Stanly Bank
MS.head()
```

```python
JPM = data.DataReader("JPM", "stooq", start, end)  # JP Morgan Bank
JPM.head()
```

```python
WFC = data.DataReader("WFC", "stooq", start, end)
WFC.head()
```

```python
GS = data.DataReader("GS", "stooq", start, end)
GS.head()
```

```python
bank_ticker = ["BAC", "C", "MS", "GS", "JPM", "WFC"]
```

```python
bank_stocks_df = pd.concat([BAC, C, MS, GS, JPM, WFC], axis=1, keys=bank_ticker)
bank_stocks_df.head()
```

```python
bank_stocks_df.columns.names = ["Bank_Ticker", "Stock_Info"]
bank_stocks_df.head()
```

```python
bank_stocks_df.xs(
    key="Close",
    axis=1,
    level="Stock_Info",
).max()
```
